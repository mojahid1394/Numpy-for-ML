```python
import numpy as np
```

sorting numpy array

```python
np1 = np.array([3,1,4,0,2,8])
print(np1)
print(np.sort(np1))
```

alphabetical

```python
np2 = np.array(["Ramim", "Mojahid", "Jhon", "Alice"])
print(np2)
print(np.sort(np2))
```

boolean

```python
np3 = np.array([True, False, True, False])
print(np3)
print(np.sort(np3))
```

2-d array sorting

```python
np4 = np.array([[6,3,8,4,1],[9,4,0,1,2]])
print(np4)
print(np.sort(np4))
```
