```python
import numpy as np
```

copy vs view

```python
np1 = np.array([0,1,2,3,4,5])
```

create view

```python
np2 = np1.view()
print(f'orginal np1 {np1}')
print(f'orginal np2 {np2}')

np1[0] = 41

print(f'changed np1 {np1}')
print(f'orginal np2 {np2}')
```

create copy

```python
np3 = np1.copy()
print(f'orginal np1 {np1}')
print(f'orginal np3 {np3}')

np1[0] = 45

print(f'changed np1 {np1}')
print(f'orginal np3 {np3}')
```
